```python
from pathlib import Path
from google.colab import files

uploaded = files.upload()

uploaded_paths = [Path(name) for name in uploaded.keys()]
image_candidates = [path for path in uploaded_paths if path.suffix.lower() in {'.jpg', '.jpeg', '.png', '.bmp', '.gif', '.webp', '.ppm'}]
video_candidates = [path for path in uploaded_paths if path.suffix.lower() in {'.mp4', '.mov', '.avi', '.mkv'}]

if not image_candidates:
    raise FileNotFoundError("Upload at least one image file.")
if not video_candidates:
    raise FileNotFoundError("Upload at least one video file.")

IMAGE_PATH = str(image_candidates[0])
VIDEO_PATH = str(video_candidates[0])
print("Using image:", IMAGE_PATH)
print("Using video:", VIDEO_PATH)
```
